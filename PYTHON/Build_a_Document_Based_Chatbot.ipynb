# Foundational Knowledge 📚

OpenAI's GPT models are renowned for their ability to provide answers across a vast array of topics. However, their knowledge is confined to what they learned up until their last training cut-off in April 2023. This limitation means they might not be able to provide information on:

- Events or developments that occurred after their last training update.
- Private documents or unpublished materials.
- Details from past interactions that weren't included in their training data.

So, what can we do when we need GPT to explore beyond its existing knowledge base?

\
**Beyond Fine-Tuning: The Power of Search**

While it might seem logical to update or 'fine-tune' GPT with new information, this approach is often impractical for extending its knowledge on specific topics or facts. Instead, we recommend leveraging search capabilities to dynamically incorporate new information into the model's responses. Here's why:

Fine-tuning is akin to long-term memory, useful for specialized tasks but not ideal for rapidly changing information. Injecting information via search acts like short-term memory, allowing GPT to access the most relevant and up-to-date data for each query.

Imagine the model as a student; fine-tuning is like cramming for an exam weeks in advance, while using search inputs is like taking an open-book test, where the student can pull in the most relevant information as needed.

\
**Limitations to Consider**

One downside of text search relative to fine-tuning is that each model is limited by a maximum amount of text it can read at once:

| Model           | Maximum text length       |
|-----------------|---------------------------|
| `gpt-3.5-turbo` | 4,096 tokens (~5 pages)   |
| `gpt-4`         | 8,192 tokens (~10 pages)  |
| `gpt-4-32k`     | 32,768 tokens (~40 pages) |

(New model is available with longer contexts, gpt-4-1106-preview have 128K context window)

Continuing the analogy, you can think of the model like a student who can only look at a few pages of notes at a time, despite potentially having shelves of textbooks to draw upon.

Therefore, to build a system capable of drawing upon large quantities of text to answer questions, we recommend using a Search-Ask approach.This can also be called Retrival Augmented Generation (RAG).

\
**Introducing Retrieval Augmented Generation (RAG)**
RAG combines the best of search technology and GPT's natural language understanding to provide answers that require knowledge beyond GPT's training data. Here's how it works in a nutshell:

1. Preparation: First, we organize the data we might want to search through, such as breaking down documents into manageable sections.
2. Search: When a query is made, we search this prepared data to find the most relevant sections.
3. Ask: We then feed these sections, along with the query, into GPT to generate an informed response.

This process allows GPT to provide answers based on a much broader dataset than it was originally trained on.

\
**Implementing Search: The Basics**

There are several ways to search through text, from simple keyword matching to more complex graph-based searches seen below:

- Lexical-based search
- Graph-based search
- Embedding-based search


This notebook uses an approach based on embeddings, which is particularly effective for matching questions with their answers due to the nuanced way embeddings capture the meaning of text. Embeddings are simple to implement and work especially well with questions, as questions often don't lexically overlap with their answers.

> **What are Embeddings?**

> Embeddings are dense vector representations of words, phrases, or even whole documents. These vectors capture semantic meanings such that words with similar meanings are close to each other in the embedding space. This is achieved by training on large text corpora, where the model learns the context of each word relative to other words.

Consider embeddings-only search as a starting point for your own system. Better search systems might combine multiple search methods, along with features like popularity, recency, user history, redundancy with prior search results, click rate data, etc. Q&A retrieval performance may also be improved with techniques like HyDE, in which questions are first transformed into hypothetical answers before being embedded. Similarly, GPT can also potentially improve search results by automatically transforming questions into sets of keywords or search terms.

\
**Full procedure of RAG**

Specifically, this notebook demonstrates the following procedure of RAG:

- Prepare search data (once per document)
- Collect: We'll download a few hundred Wikipedia articles about the 2022 Olympics
- Chunk: Documents are split into short, mostly self-contained sections to be embedded
- Embed: Each section is embedded with the OpenAI API
- Store: Embeddings are saved (for large datasets, use a vector database)
- Search (once per query)
- Given a user question, generate an embedding for the query from the OpenAI API
- Using the embeddings, rank the text sections by relevance to the query
- Ask (once per query)
- Insert the question and the most relevant sections into a message to GPT
- Return GPT's answer

\
**Costs**

It's important to note that while search operations are relatively inexpensive, generating responses with GPT, especially at higher volumes, can contribute significantly to overall costs. For instance, as of April 2023:

- A single query using GPT-3.5 Turbo costs approximately `$0.002`.
- Using GPT-4 for the same query costs around `$0.03`.


Actual costs will vary based on the specifics of your implementation and usage patterns.


### Simplifying Retrieval Augmented Generation (RAG)
Retrieval Augmented Generation (RAG) is a game-changer in the world of Large Language Models (LLMs), like GPT, by bringing your unique data into the mix. This guide will simplify the concept and stages involved in RAG to make it accessible to a wide audience, including executives and those new to this technology.

**What is RAG?**

Imagine you have a vast library at your disposal, but it doesn't contain the latest books you've just acquired. RAG is like updating that library with your new collection, ensuring that when you ask for information, you're getting the most comprehensive answers possible, including your latest additions.

In technical terms, RAG combines the vast knowledge base of pre-trained LLMs with specific, often more current or specialized data that you provide. This ensures that responses are not just based on the model's training data but are augmented with your data for more relevant and precise answers.



image.png

### The Five Key Stages of RAG

Building a system that uses RAG involves several crucial steps. Here's a breakdown of each stage in a less technical, more accessible way:

1. **Loading (Gathering Your Books)**:
This step is about collecting the data you want the LLM to consider. Think of it as gathering all the books, articles, reports, and notes you want to add to your library. This data could come from various sources, such as online articles, databases, PDFs, or even data collected from sensors.

2. **Indexing (Organizing Your Library)**:
Once you have all your data, the next step is organizing it so you can find what you need when you need it. Indexing involves creating a system (often using something called vector embeddings) that helps the LLM understand the content and context of your data. This is akin to categorizing books in your library by topics or genres so you can easily retrieve what you're looking for.

3. **Storing (Keeping Your Books on the Shelves)**:
After organizing your data, you need to store it in a way that it can be easily accessed. This involves saving your indexed data, along with any additional information that might help retrieve it later. Think of this as putting your books on the shelves, ready to be pulled out when needed.

4. **Querying (Asking Questions)**:
With your data loaded, indexed, and stored, you're now ready to ask questions. This step involves interacting with the LLM to find answers or insights based on your query. The system will search through your indexed data to find the most relevant information before responding. It's like asking a librarian for books on a specific topic, who then retrieves the best options for you.

5. **Evaluation (Making Sure Your Library Works Well)**:
Finally, it's essential to continually assess how well your system is performing. This means evaluating how accurately and quickly it responds to queries and making adjustments as necessary. It's similar to ensuring that your library's cataloging system effectively helps people find what they're looking for.

```python
#@title Let's load the data and build out an embedding Index

#@markdown Press ▶
documents = SimpleDirectoryReader("data").load_data()
index = VectorStoreIndex.from_documents(documents)
```

```python
#@title Query, Customize and talk to the LLM on that data!

#@markdown > 💡 `Model Choice`: Keep in mind, gpt-4 costs more per token!
model_choice = 'gpt-3.5-turbo' #@param ["gpt-3.5-turbo", "gpt-4"]

#@markdown \
#@markdown > 💡 `Temperature`: changes how imaginative the model can be.
temperature_choice = "0.1" #@param ["0.1", "0.5", "1.0"]
temperature_choice = float(temperature_choice)

#@markdown \
#@markdown > 💡 `Initial Prompt`: Change how the model acts when formulating its response. For example, you can instruct the model to respond in Spanish, summarize a concept, or generate ideas in a creative manner. Crafting a thoughtful initial prompt is a key part of prompt engineering. Consider specifying the style, tone, or specific details you want the model to include in its response. This helps in obtaining more targeted and useful outputs from the model.
initial_prompt = "" #@param {type:"string"}

#@markdown \
#@markdown > 💡 `Prompt`: Ask the model a question!
#@markdown >> Prompt engineering is a crucial technique in natural language processing, especially with large models like GPT-3. It involves designing prompts that guide the model to generate the desired outputs efficiently. Here are some tips:
#@markdown >> - **Precision in Prompting**: Specificity in prompts leads to more accurate responses.
#@markdown >> - **Iterative Refinement**: Refine prompts based on prior outputs to improve results.
#@markdown >> - **Understanding Capabilities**: Knowing what the model does best can help tailor your prompts.
#@markdown >> - **Use of Instructions**: Direct the model's responses by explicitly stating how it should respond (e.g., list, describe, explain).

question = "" #@param {type:"string"}

#@markdown Press ▶

if question != "":

  # Configuration for the LLM
  Settings.llm = OpenAI(temperature=temperature_choice, model=model_choice)
  query_engine = index.as_query_engine()

  # Prepare prompt
  prompt_string = initial_prompt + ". " + question if initial_prompt else question

  display(Markdown(f"<i>AI is thinking...</i>"))
  # Query and display response
  response = query_engine.query(prompt_string)
  display(Markdown(f"<b>AI RESPONSE:</b> {response}"))

else:
  print("Please provide a question and rerun the cell.")
```
