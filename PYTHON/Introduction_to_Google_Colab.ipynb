# Introduction to Google Colab Notebooks

It's especially suited for machine learning, data analysis, and education. More technically, Colab is a hosted Jupyter notebook service that requires no setup to use, while providing free access to computing resources including GPUs.

## Connecting to a Colab Instance

When you open a notebook in Google Colab, it runs on a temporary cloud server. To execute code, you first need to connect to a Colab runtime instance. Here's how you can do that:

### Starting Up

1. **Open your notebook:** Navigate to the Google Colab website and open your notebook.

2. **Connect to a runtime:** Look for the **Connect** button in the top right corner of the notebook interface. It might also display as a **Disconnected** button if you're not connected yet.

image.png

3. **Click on the Connect button:** By clicking it, Colab will start the process of allocating resources for your notebook. You'll see the status change from **Allocating** to **Connected**. This indicates that your notebook is now linked to a cloud-based runtime and you can start executing code.

    - **Resource Allocation:** Depending on the current load and your usage history, Colab will allocate a CPU, GPU, or TPU instance for your session. For GPU or TPU access, you can select the desired hardware accelerator through the notebook settings (under the 'Runtime' menu, choose 'Change runtime type').

4. **Indicator Icons:** Once connected, you'll notice a few icons next to the Connect button. These represent the health of your connection and the resources being used. Hovering over these icons provides more information about your current session's resource consumption and connection status.

### Important Tips

- **Session Timeout:** Remember, Colab sessions are temporary. After a period of inactivity (typically 90 minutes), or if you exceed the maximum allowed runtime, your session will be automatically disconnected, and any unsaved work in your environment (variables, imported libraries, etc.) will be lost.
- **Saving Work:** To prevent data loss, regularly save your work. Although Colab autosaves notebooks to Google Drive, consider manually saving important changes or downloading the notebook to your local machine.

By following these steps, you can ensure that your Colab notebooks run smoothly and efficiently, allowing you to focus on your data analysis or machine learning projects without worrying about the underlying infrastructure.

```python
import time

# If you are interested, change the number next to "a = " and see how the output changes!
a = 10

print("Let's Count!")
print("------------")
for i in range(a):
  print(f"Count:{i}")
  time.sleep(1)

print("Done!")
```
